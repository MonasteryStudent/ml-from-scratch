## Exercise 1 – Temporal-Difference State-Value Updates

Consider a continuing task with two states and a single available action. The agent follows a fixed deterministic policy and repeatedly moves between the two states:

$$
s_0 \xrightarrow{r=1} s_1,
$$

$$
s_1 \xrightarrow{r=0} s_0.
$$

The initial state-value estimates are

$$
v_0(s_0)=v_0(s_1)=0.
$$

Use the discount factor

$$
\gamma=0.5
$$

and the step size

$$
\alpha=0.5.
$$

For every observed transition, calculate the TD target

$$
\bar{v}_t
=
r_{t+1}
+
\gamma v_t(s_{t+1}),
$$

the TD error

$$
\delta_t
=
v_t(s_t)-\bar{v}_t,
$$

and the updated state-value estimate

$$
v_{t+1}(s_t)
=
v_t(s_t)-\alpha\delta_t.
$$

Calculate the first four TD updates by hand. Remember that every update uses the value estimates available at that particular time step.

### Reflection

1. Why is only the value estimate of the visited state updated?

2. Where does bootstrapping occur in the TD update?

3. Why does the reward received during the transition from $s_0$ to $s_1$ later influence the value estimate of $s_1$?

4. Why does a negative TD error increase the state-value estimate under Zhao's sign convention?

5. How would a larger step size affect the individual updates?

<details>
<summary><strong>Show reflection answers</strong></summary>

1. Only the visited state is updated because the observed transition provides new information about the value of that particular state. No new transition was observed for the other state during the same time step.

2. Bootstrapping occurs when the current estimate of the next state's value,

   $$
   v_t(s_{t+1}),
   $$

   is used to construct the TD target for the current state.

3. The first transition increases the estimate of $v(s_0)$. During the following transition from $s_1$ to $s_0$, this updated estimate becomes part of the TD target for $s_1$. Information about the reward is therefore propagated between states through bootstrapping.

4. Zhao defines the TD error as

   $$
   \delta_t=v_t(s_t)-\bar{v}_t.
   $$

   A negative TD error means that the current estimate is smaller than the TD target. Since the update subtracts the TD error,

   $$
   v_{t+1}(s_t)=v_t(s_t)-\alpha\delta_t,
   $$

   subtracting a negative value increases the estimate.

5. A larger step size moves the estimate farther toward the TD target during each update. This can accelerate learning, but it also makes the estimates more sensitive to individual transitions and stochastic variation.

</details>

## Exercise 2 – Sarsa Action-Value Updates

Consider an episodic task with two non-terminal states, two available actions, and one terminal state.

During one episode, the agent generates the following sequence:

$$
s_0
\xrightarrow{a_1,\ r=1}
s_1
\xrightarrow{a_0,\ r=2}
s_{\mathrm{terminal}}.
$$

The corresponding Sarsa experience is

$$
(s_0,a_1,1,s_1,a_0,2,s_{\mathrm{terminal}}).
$$

The initial action-value estimates are

$$
q_0(s,a)=0
$$

for every state-action pair.

Use the discount factor

$$
\gamma=0.5,
$$

the step size

$$
\alpha=0.5,
$$

and the exploration parameter

$$
\varepsilon=0.2.
$$

For each transition, calculate the Sarsa TD target

$$
\bar{q}_t
=
r_{t+1}
+
\gamma q_t(s_{t+1},a_{t+1}),
$$

the TD error

$$
\delta_t
=
q_t(s_t,a_t)-\bar{q}_t,
$$

and the updated action value

$$
q_{t+1}(s_t,a_t)
=
q_t(s_t,a_t)-\alpha\delta_t.
$$

When the next state is terminal, use

$$
q_t(s_{\mathrm{terminal}},a)=0.
$$

Complete the following table:

| $t$ | $(s_t,a_t)$ | $r_{t+1}$ | $q_t(s_t,a_t)$ | $q_t(s_{t+1},a_{t+1})$ | $\bar{q}_t$ | $\delta_t$ | $q_{t+1}(s_t,a_t)$ |

### Epsilon-greedy policy

With two actions and $\varepsilon=0.2$, every action receives the exploration probability

$$
\frac{\varepsilon}{2}
=
0.1.
$$

The greedy action additionally receives

$$
1-\varepsilon
=
0.8.
$$

Its total probability is therefore

$$
0.1+0.8=0.9.
$$

### Reflection

1. What is the main difference between the quantities estimated by TD Basic and Sarsa?

2. Why must the next action $a_{t+1}$ be selected before the Sarsa update is calculated?

3. Why is the next action value set to zero when the agent reaches the terminal state?

4. Why does the reward received at the end of this episode not immediately become part of the first action-value update?

5. Why does Sarsa use an $\varepsilon$-greedy policy instead of always selecting the greedy action?

<details>
<summary><strong>Show reflection answers</strong></summary>

1. TD Basic estimates state values,

   $$
   v_\pi(s),
   $$

   whereas Sarsa estimates action values,

   $$
   q_\pi(s,a).
   $$

   Sarsa therefore evaluates the expected return of selecting a particular action in a particular state and then continuing according to the policy.

2. The Sarsa TD target contains the action value

   $$
   q_t(s_{t+1},a_{t+1}).
   $$

   The next action must therefore be selected before the target can be calculated. Because this action is selected according to the current policy, Sarsa learns about the same policy that generates its experience.

3. No further rewards are received after an episode terminates. The expected future return from the terminal state is therefore zero, so the Sarsa target for the final transition contains only the immediate reward.

4. When the first update is performed, the action value

   $$
   q_0(s_1,a_0)
   $$

   is still zero. It is updated only during the following transition. The information about the final reward can influence earlier action values during later episodes or later visits through bootstrapping.

5. A purely greedy policy would always select the action currently considered best. It might therefore stop exploring before the action-value estimates are accurate. An $\varepsilon$-greedy policy usually selects a greedy action but retains a nonzero probability of trying every other action.

</details>
