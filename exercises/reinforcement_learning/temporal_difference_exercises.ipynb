$$
s_0 \xrightarrow{r=1} s_1,

$$
s_1 \xrightarrow{r=0} s_0.
$$

The initial state-value estimates are

$$
v_0(s_0)=v_0(s_1)=0.
$$

Use the discount factor

$$
\gamma=0.5
$$

and the step size

$$
\alpha=0.5.
$$

$$
\bar{v}_t
=
r_{t+1}
+
\gamma v_t(s_{t+1}),
$$

$$
\delta_t
=
v_t(s_t)-\bar{v}_t,

$$
v_{t+1}(s_t)

v_t(s_t)-\alpha\delta_t.

### Reflection

1. Why is only the value estimate of the visited state updated?

2. Where does bootstrapping occur in the TD update?

3. Why does the reward received during the transition from $s_0$ to $s_1$ later influence the value estimate of $s_1$?

4. Why does a negative TD error increase the state-value estimate under Zhao's sign convention?

5. How would a larger step size affect the individual updates?

<details>
<summary><strong>Show reflection answers</strong></summary>

1. Only the visited state is updated because the observed transition provides new information about the value of that particular state. No new transition was observed for the other state during the same time step.

2. Bootstrapping occurs when the current estimate of the next state's value,

   $$
   v_t(s_{t+1}),
   $$

   is used to construct the TD target for the current state.

3. The first transition increases the estimate of $v(s_0)$. During the following transition from $s_1$ to $s_0$, this updated estimate becomes part of the TD target for $s_1$. Information about the reward is therefore propagated between states through bootstrapping.

4. Zhao defines the TD error as

   $$
   \delta_t=v_t(s_t)-\bar{v}_t.
   $$

   A negative TD error means that the current estimate is smaller than the TD target. Since the update subtracts the TD error,

   $$
   v_{t+1}(s_t)=v_t(s_t)-\alpha\delta_t,
   $$

   subtracting a negative value increases the estimate.

5. A larger step size moves the estimate farther toward the TD target during each update. This can accelerate learning, but it also makes the estimates more sensitive to individual transitions and stochastic variation.

</details>
