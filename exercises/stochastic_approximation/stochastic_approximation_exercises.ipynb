## 1. Incremental Mean Estimation

### Exercise

Consider the samples

$$
x_1=2,\qquad x_2=4,\qquad x_3=3.
$$

The initial estimate is

$$
w_0=0.
$$

The incremental mean update is

$$
w_k
=
w_{k-1}
-
\alpha_k
\left(w_{k-1}-x_k\right).
$$

For each sample, calculate the step size

$$
\alpha_k=\frac{1}{k}
$$

and the resulting estimate $w_k$.

Complete the following table by hand:

| $k$ | $x_k$ | $\alpha_k$ | $w_{k-1}-x_k$ | $w_k$ |
|---:|---:|---:|---:|---:|
| 1 | 2 |  |  |  |
| 2 | 4 |  |  |  |
| 3 | 3 |  |  |  |

<summary>Show solution</summary>

$$
w_1

For the second sample,

$$
w_2

For the third sample,

$$

3-\frac{1}{3}(3-3)

The estimate history is therefore

$$

The final estimate agrees with the arithmetic mean:

$$
\frac{2+4+3}{3}=3.
$$

During the first update, the step size is $\alpha_1=1$. Therefore,

$$
w_1
=
w_0-(w_0-x_1)
=
x_1.
$$

The first sample completely replaces the initial estimate.

</details>

### Reflection

1. Why does the first estimate $w_1$ equal the first sample $x_1$?

2. How does the step size affect the influence of a new sample?

3. Why does incremental mean estimation not require storing all previous samples?

4. Where is the same averaging principle used in Monte Carlo reinforcement learning?

<details>
<summary>Show reflection answers</summary>

1. For the first update, $\alpha_1=1$. Therefore,

   $$
   w_1=w_0-(w_0-x_1)=x_1,
   $$

   so the initial estimate is completely replaced.

2. Since $\alpha_k=1/k$, the step size decreases as more samples are processed. A new sample therefore has a strong influence when $k$ is small and a weaker influence when $k$ is large.

3. Each update only requires the current estimate $w_{k-1}$, the new sample $x_k$, and the sample count $k$.

4. Monte Carlo methods estimate state or action values by averaging observed returns. These estimates can be updated incrementally without storing all previous returns.

</details>

## 2. Robbins–Monro Algorithm

### Exercise

Consider the function

$$
g(w)=w-2,
$$

whose root is

$$
w^*=2.
$$

Start with the initial estimate

$$
w_0=0
$$

and use the decreasing step size

$$
\alpha_k=\frac{1}{k+1}.
$$

For this exercise, the Robbins–Monro update is written as

$$
w_k
=
w_{k-1}
-
\alpha_k g(w_{k-1}).
$$

Calculate the first three updates by hand and complete the following table:

| $k$ | $w_{k-1}$ | $\alpha_k$ | $g(w_{k-1})$ | $w_k$ |
|---:|---:|---:|---:|---:|
| 1 | $0$ |  |  |  |
| 2 |  |  |  |  |
| 3 |  |  |  |  |

<details>
<summary>Show solution</summary>

For the first update,

$$
\alpha_1=\frac{1}{2},
\qquad
g(w_0)=0-2=-2,
$$

and therefore

$$
w_1
=
0-\frac{1}{2}(-2)
=
1.
$$

For the second update,

$$
\alpha_2=\frac{1}{3},
\qquad
g(w_1)=1-2=-1,
$$

and therefore

$$
w_2
=
1-\frac{1}{3}(-1)
=
\frac{4}{3}.
$$

For the third update,

$$
\alpha_3=\frac{1}{4},
\qquad
g(w_2)=\frac{4}{3}-2=-\frac{2}{3},
$$

and therefore

$$
w_3
=
\frac{4}{3}
-
\frac{1}{4}
\left(-\frac{2}{3}\right)
=
\frac{3}{2}.
$$

The estimates move from $w_0=0$ toward the root $w^*=2$:

$$
0,\quad 1,\quad \frac{4}{3},\quad \frac{3}{2}.
$$

</details>

### Reflection

1. Why do the updates move the estimate toward the root $w^*=2$?

2. Why are decreasing step sizes useful when the observations contain noise?

3. Does every noisy Robbins–Monro update necessarily move the estimate closer to the root?

<details>
<summary>Show reflection answers</summary>

1. When $w<2$, the function value $g(w)=w-2$ is negative. Subtracting this negative value increases $w$ and moves the estimate toward the root. When $w>2$, the function value is positive and the update decreases $w$.

2. Decreasing step sizes reduce the influence of later observation noise while still allowing the estimate to continue approaching the root.

3. No. An individual noisy observation may move the estimate away from the root. Under suitable conditions, the zero-mean noise tends to cancel over many updates and the estimates converge toward the root.

</details>

## 3. Stochastic Gradient Descent

### Exercise

Consider the sample loss

$$
f(w,x_k)=\frac{1}{2}(w-x_k)^2.
$$

Its stochastic gradient with respect to $w$ is

$$
\nabla_w f(w,x_k)=w-x_k.
$$

Use the samples

$$
x_1=2,\qquad x_2=4,\qquad x_3=3
$$

and the initial parameter

$$
w_0=0.
$$

For each iteration, use the step size

$$
\alpha_k=\frac{1}{k}
$$

and apply the SGD update

$$
w_k
=
w_{k-1}
-
\alpha_k
\nabla_w f(w_{k-1},x_k).
$$

| 2 | 4 |  |  |  |  |

$$

For the second sample,

\nabla_w f(w_1,x_2)=2-4=-2,

$$
w_2

=
3.
$$

For the third sample,

\nabla_w f(w_2,x_3)=3-3=0,

w_3

=
3.
$$

The parameter history is

$$

$$

</details>

### Reflection

1. Why does this SGD exercise produce the same updates as the incremental mean exercise?

2. Why is the gradient called stochastic?

3. Why can stochastic gradient descent be considered a special case of the Robbins–Monro algorithm?

<details>
<summary>Show reflection answers</summary>

1. The stochastic gradient is $w-x_k$. With $\alpha_k=1/k$, the SGD update is identical to the incremental mean update.

2. Each update uses the gradient calculated from a randomly sampled observation rather than the exact gradient of the expected loss.

3. Robbins–Monro searches for a root of a function using noisy observations. SGD searches for a parameter $w^*$ satisfying

   $$
   \nabla_w J(w^*)=0
   $$

   and uses the stochastic gradient

   $$
   \nabla_w f(w_k,x_k)
   $$

   as a noisy observation of the expected gradient $\nabla_w J(w_k)$. The SGD update is therefore a Robbins–Monro update applied to the gradient of an objective function.

</details>
